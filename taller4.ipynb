```python
import requests
import pandas as pd 

base_url = "https://api-colombia.com/"
endpoint = "api/v1/TouristicAttraction"
response1 = requests.get(base_url + endpoint)

print(response1.status_code)

respuesta_json = response1.json()
```

```python
data = response1.json()
print("Tipos de datos")
print(type(data))

print("Número de item en la lista de datos")
print(len(data))
```

```python
if data:
  first_item = data[0]
  print("Tipo de datos del primer item:")
  print(type(first_item))

if isinstance(first_item, dict):
    print("Llaves del diccionario:")
    print(first_item.keys())

    if 'name' in first_item:
        print("\nNombre del atractivo turístico:")
        print(first_item['name'])

    else:
        print("La llave 'name' no fue encontrada.")

    if 'description' in first_item:
        print("\nDescripción:")
        print(first_item['description'])
    else:
        print("La llave 'description' no fue encontrada.")
```

```python
df_api = pd.DataFrame.from_dict(respuesta_json)
df_api
```
